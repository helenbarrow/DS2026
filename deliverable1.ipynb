### The Simulation Itself
**Stopping Rule**
a player will use a strategy & play until:
1. the player wins **W** dollars
2. the player goes bankrupt
3. the player completes **L** plays

**Budget**
- the player starts with **B** dollars. 

**Max Wager**
- some casinos have a maximum bet,
- this parameter is called **M**

```python
# B --> starting budget --> $200
# W --> goal winnings --> $300 (B + $100 winnings)
# L --> time threshold for stopping/num plays --> 200 plays
# M --> casino's max wager --> $100

import numpy as np 
from itertools import groupby
import matplotlib.pyplot as plt 
import pandas as pandas

def oneplay(B=200, W=300, L=200, M=100, summary=True):

    def wager(ledger):
        if ledger["outcome"][-1] == 1:
            new_wager = 1
        else:
            new_wager = min(2*ledger["wager"][-1], M, ledger["budget"][-1])
        ledger["wedger"].append(new_wager)
    
    def spin(ledger):
        spin = np.random.binomial(1, p = 18/38)
        ledger["outcome"].append(spin)
    
    def stop_rule(ledger):
        if ledger["outcome"][-1] == 1:
            ledger["budget"].append(ledger["budget"][-1] + ledger["wager"][-1])
        else:
            ledger["budget"].append(ledger["budget"][-1] - ledger["wager"][-1])

        if ledger["budget"][-1] >= W:
            outcome = 1
        elif len(ledger["budget"]) - 1 >= L:
            outcome = -1
        elif ledger["budget"][-1] <=0:
            outcome = -2
        else:
            outcome = 0
        return outcome

    ledger = {"budget":[B, B], "wager":[0], "outcome":[1]}
    outcome = 0
    while outcome == 0:
        wager(ledger)
        spin(ledger)
        outome = stop_rule(ledger)
    
    if summary == False:
        return ledger
    
    bad_luck = max([len(list(values)) for key, values in groupby(ledger["outcome"]) if key == 0])

    summary = {
        "B":B,
        "W":W,
        "L":L,
        "M":M,
        "outcome":outcome,
        "profit":ledger["budget"][-1] - B,
        "n_games":len(ledger["budget"]) -1,
        "prop_red":np.mean(ledger["outcome"]),
        "bad_luck_streak": bad_luck
    }

    return summary
```
