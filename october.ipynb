```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns

from ipywidgets import interact
import ipywidgets as widgets
from IPython.display import display
```

```python
temps =  pd.read_csv("oct31-temps.csv")
temps.head()
```

```python
@interact(bins=(1, 100,1))
def plot_histogram(bins=25):
    plt.figure(figsize=(8, 4))
    temps.high.plot.hist(bins = bins, 
                     edgecolor='lightblue', 
                     linewidth = 1.2, 
                     density = True)
    plt.xlabel("temperature (Fahrenheit)")
    plt.ylabel("density")
    plt.title(f"october 31 high temperatures with {bins} bins")
    plt.show()
```

```python
plt.ecdf(temps["high"])
plt.xlabel("high temperature (fahrenheit)")
plt.ylabel("cumulative proportion")
plt.title("ECDF of cct 31 high temps")
plt.show()
```
